# CHSH Game with Qiskit
 
 This notebook explores the CHSH game, a two-player nonlocal game involving Alice and Bob. 
 It compares a deterministic classical strategy with the standard quantum strategy based on a shared entangled state.

## Game Rule
 
 Alice receives a bit `x`, Bob receives a bit `y`, and they return bits `a` and `b` without communicating. 
 They win when:
 
 ```text
 a XOR b = x AND y
 ```


## Quantum Circuit
 
 The quantum strategy starts by preparing an entangled pair. Alice and Bob then choose measurement angles based on their input bits.


````python
build_chsh_circuit(0, 0).draw(output="mpl")
````

## Input-Dependent Circuits
 
 These are the four possible circuits, one for each pair of input bits.


````python
for x in (0, 1):
     for y in (0, 1):
         print(f"(x, y) = ({x}, {y})")
         display(build_chsh_circuit(x, y).draw(output="mpl"))
````

## Quantum Strategy
 
 The quantum strategy runs the input-dependent circuit once and uses the measured bits as Alice and Bob's answers.


````python
NUM_GAMES = 1000
 quantum_rate = estimate_win_rate(quantum_strategy, games=NUM_GAMES, seed=7)
 print(f"Quantum strategy win rate: {quantum_rate:.3f}")
````

## Classical Baseline
 
 A deterministic classical strategy can win at most 75% of CHSH rounds.


````python
classical_rate = estimate_win_rate(classical_strategy, games=NUM_GAMES, seed=7)
 print(f"Classical strategy win rate: {classical_rate:.3f}")
````

````python
print(f"Quantum advantage over this classical baseline: {quantum_rate - classical_rate:.3f}")
````
